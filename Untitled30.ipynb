# 収益改善のためのデータ分析

## データの読み込みと概要の確認

まず、`fraud.csv` ファイルをPandas DataFrameに読み込み、データの構造と最初の数行を確認します。これにより、どのようなデータが含まれているかを把握し、その後の分析の方向性を定めます。

```python
import pandas as pd

# fraud.csvファイルを読み込む
df = pd.read_csv('/content/fraud.csv')

# データの最初の5行を表示
display(df.head())
```

## データ型の確認と欠損値の特定

データの品質を評価し、適切な前処理計画を立てるために、各列のデータ型（`dtypes`）を確認し、欠損値（`NaN`）の数を特定します。これにより、データがどのように構成されているか、そしてどの列にデータクリーンアップが必要かを把握できます。

```python
print('データフレーム情報:')
df.info()

print('\n欠損値の数:')
print(df.isnull().sum())
```

## 欠損値の処理：メディアンによる補完

データの多くの列に欠損値が存在することが確認されました。これらの欠損値をそのままにしておくと、その後の分析の精度や結果に悪影響を及ぼす可能性があります。ここでは、数値型の列の欠損値を、その列の**メディアン（中央値）**で補完します。メディアンは外れ値の影響を受けにくいため、平均値よりも堅牢な補完方法として選択しました。

欠損値が多い`velocity_score`や`customer_age`、`distance_from_home`、`transaction_amount`、`network_quality`などに特に注意して処理を行います。

```python
# 数値型の列を特定し、メディアンで欠損値を補完する
for column in df.select_dtypes(include=['float64', 'int64']).columns:
    if df[column].isnull().any():
        median_value = df[column].median()
        df[column].fillna(median_value, inplace=True)
        print(f"列 '{column}' の欠損値をメディアン({median_value})で補完しました。")

# 欠損値が全て処理されたか再確認
print('\n欠損値処理後の欠損値の数:')
print(df.isnull().sum())
```

## データクリーニングの完了と詐欺トランザクションの分布確認

欠損値の処理が完了し、データセットが分析に適した状態になりました。次に、主要な目的変数である`is_fraud`（詐欺かどうか）の分布を確認します。詐欺トランザクションの割合を把握することは、収益改善のための課題特定において非常に重要です。このデータセットにおける詐欺の発生率を理解し、その後の分析の出発点とします。

```python
# is_fraudカラムの分布を確認
fraud_counts = df['is_fraud'].value_counts()
fraud_percentage = df['is_fraud'].value_counts(normalize=True) * 100

print("is_fraudカラムの分布:")
print(fraud_counts)
print("\nis_fraudカラムの割合 (%):")
print(fraud_percentage)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6, 4))
sns.barplot(x=fraud_counts.index, y=fraud_counts.values, palette='viridis')
plt.title('詐欺取引の分布 (0: 詐欺ではない, 1: 詐欺)')
plt.xlabel('is_fraud')
plt.ylabel('件数')
plt.xticks([0, 1], ['詐欺ではない', '詐欺'])
plt.show()
```

## データの特徴量概要と詐欺取引による違い

詐欺取引の分布を確認したところで、次に各数値特徴量の統計的な概要を見ていきます。特に、詐欺取引と非詐欺取引の間で各特徴量にどのような違いがあるのかを比較することで、詐欺を特定するための手がかりや、収益に影響を与える要因を洗い出すことができます。これにより、どのような行動や取引パターンが収益を減少させているのか、仮説を立てるための基礎を築きます。

```python
# 詐欺ではない取引と詐欺取引に分けて記述統計量を比較
print("\n--- 詐欺ではない取引 (is_fraud = 0) の記述統計量 ---")
display(df[df['is_fraud'] == 0].describe())

print("\n--- 詐欺取引 (is_fraud = 1) の記述統計量 ---")
display(df[df['is_fraud'] == 1].describe())
```

## 課題定義

現在のデータ分析から、全体のトランザクションの**10.3%が詐欺取引**であることが判明しました。この高い詐欺取引率は、企業にとって直接的な金銭的損失（チャージバックなど）だけでなく、顧客からの信頼低下、不正検知システムへの追加投資、および運用の手間など、多大なコストと機会損失を引き起こしています。

**主な課題:** 高い詐欺取引率が企業の収益性を圧迫しており、これを効果的に削減する必要がある。

## 仮説

詐欺取引と非詐欺取引の間には、特定の取引特徴量（例：取引金額、顧客年齢、場所からの距離、デバイスの種類、取引の履歴など）において明確な違いが存在する可能性があります。これらの違いを特定し、不正検知の強化やビジネスプロセスの改善に活用することで、詐欺取引率を低減し、結果として収益を最大化できると仮説を立てます。

```python
# 全特徴量間の相関行列を計算
correlation_matrix = df.corr()

# 'is_fraud'との相関が高い特徴量を抽出
fraud_correlation = correlation_matrix['is_fraud'].sort_values(ascending=False)

# 相関ヒートマップの作成
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix[['is_fraud']], annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('各特徴量とis_fraudの相関ヒートマップ')
plt.show()

print("\nis_fraudとの相関ランキング:")
print(fraud_correlation)
```

線形相関が低い場合でも、非線形な関係性や特定の閾値を超えた場合に詐欺の傾向が強まることがあります。ここでは、`is_fraud`との相関が比較的高い、またはビジネス上重要と考えられる特徴量について、散布図を用いて非線形な関係を視覚的に分析します。

特に、`distance_from_home`、`transaction_amount`、`device_type`に焦点を当てて可視化を行います。`is_fraud`がカテゴリカル変数であるため、データ点の重なりを防ぐためにジッターを適用します。

## 分析: `device_type`と`is_fraud`のクロス集計

`device_type`と`is_fraud`の散布図では、特定の傾向が読み取りにくい部分がありました。そこで、クロス集計表を作成し、各`device_type`が詐欺取引と非詐欺取引にどのように分布しているか、具体的な件数と割合で確認します。これにより、不正利用されやすいデバイスタイプを特定しやすくなります。

```python
# device_type と is_fraud のクロス集計表を作成
crosstab_device_fraud = pd.crosstab(df['device_type'], df['is_fraud'])
display(crosstab_device_fraud)

# 各device_typeにおける詐欺の割合を計算
crosstab_device_fraud_percentage = pd.crosstab(df['device_type'], df['is_fraud'], normalize='index') * 100
crosstab_device_fraud_percentage.columns = ['非詐欺 (%)', '詐欺 (%)']
display(crosstab_device_fraud_percentage)

# 詐欺取引の割合でソートして表示
print("\n各デバイスタイプにおける詐欺取引の割合 (昇順):")
display(crosstab_device_fraud_percentage.sort_values(by='詐欺 (%)', ascending=False))
```

## 分析: `device_type` 0 における`transaction_amount`の行動特徴

前の分析で`device_type` 0 が最も詐欺取引の割合が高いことが判明しました。ここでは、この特定のデバイスタイプに焦点を当て、詐欺取引と非詐欺取引の間で`transaction_amount`の分布にどのような違いがあるのかを深掘りします。これにより、`device_type` 0 を通じた詐欺の具体的な行動パターンを理解し、よりターゲットを絞った対策を検討するための情報とします。

```python
import matplotlib.pyplot as plt
import seaborn as sns

# device_type が 0 のデータのみを抽出
df_device_0 = df[df['device_type'] == 0]

plt.figure(figsize=(10, 6))
sns.histplot(df_device_0[df_device_0['is_fraud'] == 0]['transaction_amount'], color='blue', label='非詐欺取引 (Device 0)', kde=True, alpha=0.5)
sns.histplot(df_device_0[df_device_0['is_fraud'] == 1]['transaction_amount'], color='red', label='詐欺取引 (Device 0)', kde=True, alpha=0.5)
plt.title('device_type 0 における取引金額の分布 (詐欺 vs 非詐欺)')
plt.xlabel('取引金額')
plt.ylabel('度数')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## 特徴量エンジニアリング: 詐欺との相関が高い新たな指標の作成

これまでの分析で得られた洞察に基づき、詐欺 (`is_fraud`) との関連性が高まるような新たな特徴量を作成します。特に、`distance_from_home`の異常値と、`device_type` 0 における低額取引という2つのパターンに注目します。

```python
import numpy as np

# 1. `abs_distance_from_home_deviation` の作成
# distance_from_home のメディアンを計算
median_distance = df['distance_from_home'].median()
# メディアンからの絶対偏差を計算
df['abs_distance_from_home_deviation'] = np.abs(df['distance_from_home'] - median_distance)

# 2. `is_device_0_low_amount` の作成
# device_type 0 かつ詐欺取引の transaction_amount のメディアンを閾値とする
# df_device_0 は前のセルで作成済み
fraudulent_amount_on_device_0_median = df_device_0[df_device_0['is_fraud'] == 1]['transaction_amount'].median()

# 新しい特徴量を作成: device_typeが0で、かつ取引金額が上記のメディアン以下であるか
df['is_device_0_low_amount'] = ((df['device_type'] == 0) & (df['transaction_amount'] <= fraudulent_amount_on_device_0_median)).astype(int)

print("新しい特徴量 'abs_distance_from_home_deviation' と 'is_device_0_low_amount' を作成しました。")
display(df[['distance_from_home', 'abs_distance_from_home_deviation', 'device_type', 'transaction_amount', 'is_device_0_low_amount', 'is_fraud']].head())
```

### 新しい特徴量と`is_fraud`の相関確認

作成した新しい特徴量と`is_fraud`との相関を計算し、以前の相関ランキングと比較してみましょう。

```python
# 新しい特徴量を含めた全特徴量間の相関行列を再計算
correlation_matrix_new = df.corr()

# 'is_fraud'との相関が高い特徴量を抽出（新しい特徴量も含む）
fraud_correlation_new = correlation_matrix_new['is_fraud'].sort_values(ascending=False)

print("\nis_fraudとの相関ランキング (新しい特徴量を含む):")
display(fraud_correlation_new)
```

## 提案: 詐欺取引削減と収益向上に向けた具体的なアクション

### 1. `distance_from_home`に基づく不正検知ルールの強化

**分析結果:** `distance_from_home`は詐欺取引と比較的高い相関（0.0468）を示し、特に自宅から極端に離れた取引や、異常に自宅に近い取引に詐欺の傾向が見られました。新たに作成した`abs_distance_from_home_deviation`も関連性を示しています。

**提案:**

*   **異常な距離からの取引の監視:**
    *   ユーザーの通常の行動範囲から著しく逸脱した距離（例: 50km以上、または極端に0に近い）からの取引に対して、追加認証（OTP、生体認証など）を要求する。
    *   地理情報に基づいたリスクスコアリングを導入し、高リスクな取引にフラグを立てる。
    *   過去の取引履歴と比較し、ユーザーが通常訪れない場所からの高額取引に注意を払う。

### 2. `device_type` 0と少額取引の組み合わせへの対策

**分析結果:** `device_type` 0は最も詐欺取引の割合が高く（約11.51%）、このデバイスタイプからの詐欺取引は特に少額（例: メディアン以下）に集中していることが判明しました。`is_device_0_low_amount`特徴量もその関連性を示しています。

**提案:**

*   **`device_type` 0からの少額取引に対する厳格な監視:**
    *   `device_type` 0からの特定の少額取引に対しては、リアルタイムでの追加認証を必須とするか、一時的に取引を保留し手動レビュープロセスに回す。
    *   不正アクセス防止のため、`device_type` 0からのログインに対しては、より強固なパスワードポリシーや二段階認証を推奨・強制する。
    *   `device_type` 0として識別されるデバイスが、既知の不正利用されたデバイスリストと一致しないか定期的にチェックする。

### 3. 機械学習モデルによる詐欺検知の導入・強化

**分析結果:** 各特徴量と`is_fraud`との線形相関は全体的に低いものの、複数の特徴量の組み合わせや非線形な関係性が詐欺に寄与している可能性が示唆されました。特に、今回作成した`abs_distance_from_home_deviation`や`is_device_0_low_amount`といった特徴量は、手動ルールでは発見しにくい異常を捉える可能性を秘めています。

**提案:**

*   **予測モデルの構築:**
    *   ロジスティック回帰、決定木、ランダムフォレスト、勾配ブースティングなどの機械学習アルゴリズムを用いて、詐欺取引を予測するモデルを構築する。
    *   今回作成した新特徴量 (`abs_distance_from_home_deviation`, `is_device_0_low_amount`) をモデルの入力特徴量として活用し、予測精度を向上させる。
    *   モデルの出力として詐欺の確率スコアを算出し、スコアに基づいて取引の自動承認、追加認証要求、手動レビューなどのアクションを分岐させる。
*   **継続的なモデルの改善:**
    *   新しい詐欺パターンに対応するため、モデルを定期的に再トレーニングし、最新のデータで評価を行う。
    *   誤検知（False Positive）と見逃し（False Negative）のバランスを考慮し、ビジネス要件に合わせた閾値調整を行う。

### 4. ユーザー行動プロファイルの活用

### 5. 継続的な監視とA/Bテスト

**提案:**

# 📈 詐欺取引分析による収益改善プロジェクト

## 概要

本プロジェクトは、トランザクションデータ（`fraud.csv`）を分析し、詐欺取引を特定・削減することで企業の収益向上とコスト最適化を目指します。データクリーニングから探索的データ分析、特徴量エンジニアリング、そして具体的な施策提案までの一連のプロセスを追っています。

## データセット

使用データセット: `fraud.csv`

このデータセットには、個々の取引に関する詳細情報が含まれており、以下の特徴量が含まれています。
- `transaction_amount`: 取引金額
- `hour_of_day`: 取引時刻（時間帯）
- `is_weekend`: 週末かどうか
- `num_items`: 取引アイテム数
- `customer_age`: 顧客の年齢
- `prev_transactions`: 過去の取引数
- `distance_from_home`: 自宅からの距離
- `device_type`: デバイスタイプ
- `network_quality`: ネットワーク品質
- `is_first_transaction`: 初回取引かどうか
- `store_type`: 店舗タイプ
- `velocity_score`: 取引速度スコア
- `is_fraud`: 詐欺取引かどうか (目的変数: 0=非詐欺, 1=詐欺)

## 分析の目的

1.  `fraud.csv`データセットの構造と品質を理解する。
2.  詐欺取引のパターンと非詐欺取引との違いを特定する。
3.  詐欺取引を予測するための新たな特徴量を設計する。
4.  分析結果に基づき、詐欺対策と収益向上に繋がる具体的な施策を提案する。

## 分析プロセスと結果

### 1. データ読み込みと初期確認
- `fraud.csv`をPandas DataFrameに読み込み、`df.head()`でデータ構造を確認。
- `df.info()`と`df.isnull().sum()`により、多くの数値列に欠損値が存在することを特定。

### 2. 欠損値の処理
- 数値型の列の欠損値を、外れ値に強い**メディアン（中央値）**で補完しました。これにより、データセット全体の欠損値が解消され、分析の準備が整いました。

### 3. 詐欺取引の分布確認
- `is_fraud`カラムの分布を確認した結果、**全体の10.3%が詐欺取引**であることが判明。これは企業にとって重大な損失に繋がる割合であり、削減の必要性が浮き彫りになりました。

### 4. データの特徴量概要と詐欺取引による違い
- 詐欺取引 (`is_fraud=1`) と非詐欺取引 (`is_fraud=0`) の間で各特徴量の記述統計量（平均、中央値、標準偏差など）を比較し、潜在的な違いを探索しました。

### 5. 課題定義と仮説
- **課題:** 高い詐欺取引率（10.3%）が企業の収益性を圧迫しており、効果的な削減策が必要。
- **仮説:** 詐欺取引と非詐欺取引の間には、`transaction_amount`、`distance_from_home`、`device_type`などの特徴量に明確な違いが存在し、これらを特定・活用することで詐欺取引率を低減できる。

### 6. 詐欺取引と非詐欺取引の比較分析
- **日本語フォント対応:** Matplotlibでの日本語表示問題を解決するため、`IPAPGothic`フォントを導入し、グラフの可読性を確保しました。
- **ヒストグラムによる分布比較:**
    - `transaction_amount`: 詐欺取引は**比較的少額の取引に集中**する傾向。
    - `distance_from_home`: 詐欺取引は**自宅からの距離が極端（非常に近い、または非常に遠い）な取引に多く発生**する傾向。
    - `velocity_score`: 詐欺取引でやや低い傾向が見られる可能性が示唆されました。

### 7. 各特徴量と`is_fraud`の相関関係
- 相関ヒートマップにより、各特徴量と`is_fraud`との線形相関を可視化。全体的に相関は低いものの、`distance_from_home`が最も高い正の相関を示しました（約0.047）。これは、詐欺が単一要因ではなく、複数の要因の複雑な組み合わせや非線形な関係によって発生している可能性を示唆します。

### 8. 非線形関係の探索
- `stripplot`を用いて、`distance_from_home`、`transaction_amount`、`device_type`と`is_fraud`の非線形な関係性を視覚的に探索。
    - `distance_from_home`: 非詐欺取引は自宅に近い距離に集中する一方、詐欺取引は広範囲に分布し、極端な距離での発生が目立ちました。
    - `transaction_amount`: 詐欺取引は、非詐欺取引と比較して、取引金額が低い範囲に密集していることが確認されました。

### 9. `device_type`と`is_fraud`のクロス集計
- `device_type`ごとに詐欺取引の割合を分析した結果、`device_type` 0 が最も詐欺取引の発生率が高い（約11.51%）ことを特定しました。

### 10. `device_type` 0 における`transaction_amount`の行動特徴
- `device_type` 0からの詐欺取引は、特に**少額の取引に顕著に集中**していることが明らかになりました。これは、このデバイスタイプを悪用した少額での不正利用が多発している可能性を示唆します。

### 11. 特徴量エンジニアリング
- 分析の洞察に基づき、`is_fraud`との相関を高めることを目的として以下の2つの新しい特徴量を作成しました。
    1.  `abs_distance_from_home_deviation`: `distance_from_home`のメディアンからの絶対偏差。自宅からの距離の「異常性」を捉えることを意図しています。
    2.  `is_device_0_low_amount`: `device_type`が0で、かつ`transaction_amount`が`device_type` 0における詐欺取引のメディアン以下の取引である場合に1となる二値特徴量。`device_type` 0における少額詐欺のパターンを直接的に捉えます。

### 12. 新しい特徴量と`is_fraud`の相関確認
- 新しい特徴量を含む全特徴量と`is_fraud`の相関を再計算。
    - `abs_distance_from_home_deviation`: 0.0416
    - `is_device_0_low_amount`: 0.0188
- これらの特徴量は、線形相関では元の特徴量と比較して劇的な改善はなかったものの、特定の詐欺パターンを捉えるための有用なシグナルを提供し、機械学習モデルの性能向上に貢献する可能性があります。
